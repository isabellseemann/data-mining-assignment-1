3.07 Problem 1

```python
from itertools import combinations


def find_all_bases(vectors):
    # Check if list is empty
    if not vectors:
        print("The list of vectors is empty.")
        return

    # Check all vectors have same dimension
    dims = [len(v) for v in vectors]
    if len(set(dims)) != 1:
        print("The list of vectors contains vectors of different dimensions.")
        return

    D = dims[0]  # dimension of each vector

    # If number of vectors < D, no basis possible
    if len(vectors) < D:
        print("There are less than D independent vectors in the set.")
        return

    vectors = [np.array(v) for v in vectors]

    bases = []

    # Check all combinations of vectors of size D
    for combo in combinations(vectors, D):
        matrix = np.column_stack(combo)  # form matrix with these vectors as columns
        rank = np.linalg.matrix_rank(matrix)

        if rank == D:
            bases.append(combo)

    if not bases:
        print("There are less than D independent vectors in the set.")
    else:
        print(f"Found {len(bases)} possible bases:")
        for i, base in enumerate(bases, start=1):
            print(f"Basis {i}:")
            for vec in base:
                print(vec.tolist())
            print()
```

```python
def get_parallel_component(vec1, vec2):
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)
    magnitude = np.linalg.norm(vec2)
    vec2_direction = vec2 / magnitude
    parallel_component = np.dot(vec2_direction, vec1) * vec2_direction
    return parallel_component


print(get_parallel_component([1, 2, 3, 4], [0, 0, 1, 0]))
```

```python
def get_projection(orthonormal_basis, vector):
    orthonormal_basis = np.array(orthonormal_basis, dtype=float)
    vector = np.array(vector, dtype=float)
    projection = np.zeros_like(vector)
    for basic_vec in orthonormal_basis:
        projection += get_parallel_component(vector, basic_vec)
    return projection
```
