```python
import numpy as np

L = [[1, 2, 3, 4], [5, 6, 7, 8], [6, 8, 10, 12], [4, 4, 4, 4], [3, 2, 1, 0]]
```

3.07 Problem 1: Bases

```python
from itertools import combinations


def find_all_bases(vectors: list[list[int]]) -> list[int] | None:
    if not vectors:
        print("The list of vectors is empty.")
        return

    dimensions = [len(vector) for vector in vectors]
    if len(set(dimensions)) != 1:
        print("The list of vectors contains vectors of different dimensions.")
        return

    dimension_of_each_vector = dimensions[0]

    vectors_length = len(vectors)
    if vectors_length < dimension_of_each_vector:
        print(f"There are less than {vectors_length} independent vectors in the set.")
        return

    vectors = [np.array(vector) for vector in vectors]

    bases = []

    for combination in combinations(vectors, dimension_of_each_vector):
        matrix = np.column_stack(combination)
        rank = np.linalg.matrix_rank(matrix)

        if rank == dimension_of_each_vector:
            bases.append(combination)

    if not bases:
        print(f"There are less than {vectors_length} independent vectors in the set.")
    else:
        print(f"Found {len(bases)} possible bases:")
        for i, base in enumerate(bases, start=1):
            print(f"Basis {i}:")
            for vec in base:
                print(vec.tolist())
            print()
```

```python
find_all_bases(L)
```

3.07 Problem 2: Parallel component of the projected vector

```python
def get_parallel_component(vec1: list[int], vec2: list[int]) -> np.ndarray:
    vector1 = np.array(vec1)
    vector2 = np.array(vec2)
    magnitude = np.linalg.norm(vector2)
    vector2_direction = vector2 / magnitude
    parallel_component = np.dot(vector2_direction, vector1) * vector2_direction
    return parallel_component


print(get_parallel_component([1, 2, 3, 4], [0, 0, 1, 0]))
```
