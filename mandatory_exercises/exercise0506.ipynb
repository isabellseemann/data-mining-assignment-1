# Interactive Visualization and Exploratory Data Analysis

## Generate an Exploratory Data Analysis report

```python
import plotly.express as px

from ydata_profiling import ProfileReport

print("Libraries imported successfully!")
print("YData Profiling enables automated exploratory data analysis")
print("Perfect for getting quick insights into any dataset")
```

```python
import pandas as pd
```

```python
# 1. load the dataset
url = "https://raw.githubusercontent.com/tidyverse/dplyr/main/data-raw/starwars.csv"
```

```python
df = pd.read_csv(url)
print(f"Dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print("\nFirst few rows:")
print(df.head())
print(f"\nDataset info:")
print(f"- {df.shape[0]} samples")
print(f"- {df.shape[1]} features")
print(f"- Species: {df['species'].unique()}")
```

```python
# 2. generate the profiling report
profile = ProfileReport(
    df, title="Star Wars dataset Profiling Report", explorative=True)

print("Profiling report generated successfully!")
print("Report configuration:")
print("- Title: 'Star Wars Dataset Profiling Report'")
print("- Explorative mode: True (includes advanced analysis)")
print("- Ready to save as HTML file")
```

```python
# 3. save the report
profile.to_file("Star_Wars_dataset_report.html")

print("Report saved successfully!")
print("File: Star_Wars_dataset_report.html")
print("Open this file in your web browser to view the complete analysis")
print("\nThe report includes:")
print("- Dataset overview and statistics")
print("- Variable distributions and summaries")
print("- Correlation analysis")
print("- Missing value analysis") 
print("- Data quality warnings")
print("- Sample data preview")
```

## Interactive scatter plot with Plotly

```python
import plotly.io as pio
pio.renderers.default = "iframe_connected"
```

```python
print(f"Columns: {list(df.columns)}")
```

```python
fig = px.scatter(
    df, x="height", y="mass", color="species",
    title="Interactive star wars Dataset Plot"
)
fig.show()

print("Interactive scatter plot created!")
```

## Interactive Bar Plot for Categorical Variables

```python
species_gender_count = (
    df.groupby(['species', 'gender'])
      .size()
      .reset_index(name='count')
)

species_gender_count['species_total'] = species_gender_count.groupby('species')['count'].transform('sum')
species_gender_count = species_gender_count.sort_values(by='species_total', ascending=False)

fig = px.bar(
    species_gender_count,
    x='species',
    y='count',
    color='gender',
    custom_data=['gender', 'species_total'],
    title='Interactive Bar Plot for Species and Gender Distribution'
)


fig.update_traces(
    hovertemplate=(
        'Species: %{x}<br>'
        'Gender: %{customdata[0]}<br>'
        'Gender Count: %{y}<br>'
        'Total Species Count: %{customdata[1]}'
    )
)

fig.show()
```
