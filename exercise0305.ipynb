```python
import numpy as np
```

```python
def solve_linear_equation(A, b):
    '''
    solve the system of linear equations Ax = b
    A: Coefficient matrixs
    b: constant matrixs
    return x , y ...
    '''
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    x = np.linalg.solve(A, b)
    print('solution:', x)
```

```python
A = np.array([[2,3],
            [4,1]])
b = np.array([5,
             6])
```

```python
solve_linear_equation(A , b)
```

```python
A1 = np.array([[2,5,8],
              [5,8,91],
              [36,54,8]])
b1 = np.array([2,
              8,
              1])
```

```python
solve_linear_equation(A1, b1)
```
